```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
import time

#  Set up WebDriver
driver = webdriver.Chrome()

#  Open LinkedIn login page
driver.get("https://www.linkedin.com/login")
time.sleep(3)  # ✅ Give time to load

# Enter login details
username = "Your email Address"  # Linkedin email
password = "Your password"  # LinkedIn password

email_field = driver.find_element(By.ID, "username")
email_field.send_keys(username)

password_field = driver.find_element(By.ID, "password")
password_field.send_keys(password)
password_field.send_keys(Keys.RETURN)

time.sleep(5)  # login to complete

# ✅ Now navigate to the UNT Alumni search page
search_url = "https://www.linkedin.com/search/results/people/?keywords=University%20of%20North%20Texas%20Alumni"
driver.get(search_url)

time.sleep(5)  # ✅ Wait for results to load
```

```python
from bs4 import BeautifulSoup
import time

time.sleep(5)  #  this the items number show for searching which can be changed

soup = BeautifulSoup(driver.page_source, "html.parser")

# Extract profile 
profile_links = []
for link in soup.find_all('a', href=True):  # here is finding the a tag
    href = link['href']
    if "linkedin.com/in/" in href and href not in profile_links:
        profile_links.append(href)

print(f"✅ HERE {len(profile_links)} alumni profiles")
print(profile_links[:10])  # Show 
print(driver.page_source[:1000])  # ✅ Print first 1000 characters
```

```python
# if you want more result just keep doing it
for _ in range(5):  # Scroll down 5 times
    driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
    time.sleep(3)  # ✅ Wait for content to load
```

```python
def scrape_profile(profile_url):
    driver.get(profile_url)  # Open LinkedIn profile page
    time.sleep(5)  # ✅ Wait for the page to load

    soup = BeautifulSoup(driver.page_source, "html.parser")

    # Extract Name
    name_tag = soup.find('h1')
    name = name_tag.get_text().strip() if name_tag else "Not Found"

    # Extract Job Title for the Alumni
    headline_tag = soup.find('div', {'class': 'text-body-medium'})
    headline = headline_tag.get_text().strip() if headline_tag else "Not Found"

    # Alumni Location
    location_tag = soup.find('span', {'class': 'text-body-small'})
    location = location_tag.get_text().strip() if location_tag else "Not Found"

    # Return their Data
    return {
        'name': name,
        'headline': headline,
        'location': location,
        'profile_url': profile_url
    }
```

```python
import pandas as pd

# Storing the data
all_profiles = []

for url in profile_links[:10]:  # limitation for testing
    profile_data = scrape_profile(url)
    all_profiles.append(profile_data)
    print(f"Scraped: {profile_data['name']}")  # printitng

# converting to csv files 
df = pd.DataFrame(all_profiles)
df.to_csv("UNT_Alumni_Data.csv", index=False)

print("Scrapped has been completed. Data saved to UNT_Alumni_Data.csv")
```
