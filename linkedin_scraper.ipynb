```python
def get_exp(exp):
    exp_dict = {}

    # Extract Company Name
    name_tag = exp.find('div', {'class': 'display-flex flex-wrap align-items-center full-height'})
    name = name_tag.find('span', {'class': 'visually-hidden'}) if name_tag else None
    name = name.get_text().strip() if name else "Not Found"

    # Extract Duration
    duration_tag = exp.find('span', {'class' : 't-14 t-normal'})
    duration = duration_tag.find('span', {'class' : 'visually-hidden'}) if duration_tag else None
    duration = duration.get_text().strip() if duration else "Not Found"

    exp_dict['company name'] = name
    exp_dict['duration'] = duration

    # Extract Designations
    designations = exp.find_all('div', {'class' : 'DMzFPveGnEOYALHlJgQkCNfpqinNPuoldmDbo'})
    
    item_list = []  
    for position in designations:
        spans = position.find_all('span', {'class' : 'visually-hidden'})

        # Ensure spans exist before accessing them
        if len(spans) < 3:
            continue  # Skip if there aren't enough spans

        item_dict = {}  # ✅ FIXED: Changed from [] to {}
        item_dict['designations'] = spans[0].get_text().strip()
        item_dict['duration'] = spans[1].get_text().strip()
        item_dict['location'] = spans[2].get_text().strip()

        # Extract Projects (if available)
        try:
            item_dict['projects'] = spans[3].get_text().strip()
        except IndexError:
            item_dict['projects'] = ""

        item_list.append(item_dict)  # ✅ FIXED: Moved this inside the loop

    exp_dict['designations'] = item_list

    return exp_dict
   
# Process all experiences
item_list = [get_exp(exp) for exp in experience]

# Store in profile_data
profile_data['experience'] = item_list
```

```python
def get_edu(item):
    span = item.find_all('span', {'class': 'visually-hidden'})
    item_dict = {}

    # Ensure span has enough elements before accessing them
    item_dict['college'] = span[0].get_text().strip() if len(span) > 0 else None
    item_dict['degree'] = span[1].get_text().strip() if len(span) > 1 else None
    item_dict['duration'] = span[2].get_text().strip() if len(span) > 2 else None
    item_dict['project'] = span[3].get_text().strip() if len(span) > 3 else None

    return item_dict

item_list = []
for item in items:
    item_list.append(get_edu(item))

profile_data['education'] = item_list
```
